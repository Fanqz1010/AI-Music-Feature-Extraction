```python
import os
import librosa
import pandas as pd

# 你的文件夹路径（在桌面上的 kling_samples）
folder_path = '/Users/fanqianzi/Desktop/kling_samples'

# 用来存放结果列表
results = []

# 遍历文件夹里的所有文件
for filename in sorted(os.listdir(folder_path)):
    if filename.endswith(('.mp3', '.wav')):
        file_path = os.path.join(folder_path, filename)
        print(f"正在处理：{filename} ...")
        
        try:
            # 加载音频（注意：没有加太复杂的参数，保持原有速度）
            y, sr = librosa.load(file_path)
            
            # 1. 提取 MFCC（音色/质感）
            mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)
            
            # 2. 提取频谱质心（明亮度）
            cent = librosa.feature.spectral_centroid(y=y, sr=sr)
            
            # 存入列表
            results.append({
                '文件名': filename,
                '时长(秒)': round(len(y)/sr, 2),
                '频谱质心均值': round(float(cent.mean()), 2),
                'MFCC均值_1': round(float(mfccs[0].mean()), 2),
                'MFCC均值_2': round(float(mfccs[1].mean()), 2)
            })
        except Exception as e:
            print(f"哎呀，{filename} 处理失败了，原因：{e}")

# 把结果存进 DataFrame 表格
df = pd.DataFrame(results)
print("\n所有音频处理完成！")
df
```
